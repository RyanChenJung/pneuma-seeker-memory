```python
import os
import sys

sys.path.append("../..")

from pneuma_seeker.utils.logger import setup_logger
from pneuma_seeker.model.interface.impl.azure import AzureOpenAILLM
from pneuma_seeker.model.llm_message import LLMMessage, Role
from pneuma_seeker.model.option import LLMOption
from pneuma_seeker.utils.config import Config

logger = setup_logger(log_path=os.path.join(".", "log"))
azure = AzureOpenAILLM("o4-mini", Config("../../../.env"), logger)
```

```python
llm_messages = [
    LLMMessage(role=Role.USER.value, content="Hi, how are you?"),
]
```

```python
final_ans = ""
gen = azure.chat(llm_messages, LLMOption(stream=True))
for log_message in gen:
    final_ans += log_message
```
